```python
from keras.models import Sequential
from keras.layers import Dense,Flatten,Conv2D,MaxPooling2D
from keras.preprocessing.image import ImageDataGenerator

#Data
data='./Weather_1'

#Resize all images
data_generator=ImageDataGenerator(rescale=1./255,validation_split=0.3)
train_gen=data_generator.flow_from_directory(data,target_size=(128,128),batch_size=32,class_mode='categorical',shuffle=True,subset='training')
valid_gen=data_generator.flow_from_directory(data,target_size=(128,128),batch_size=32,class_mode='categorical',shuffle=True,subset='validation')

classes=train_gen.num_classes

#Building CNN
model=Sequential()
model.add(Conv2D(64, (3,3), input_shape=(128,128,3), padding='same', strides=(1,1),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),trainable=False))

model.add(Conv2D(64, (3,3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2),trainable=False))

#ANN
model.add(Flatten())
model.add(Dense(32,activation='relu'))
model.add(Dense(classes,activation='softmax'))

#Compiling model
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])

#Fit model
model.fit(train_gen,epochs=15,validation_data=valid_gen)

print(model.summary())

model.save('./sys1.keras')
```
